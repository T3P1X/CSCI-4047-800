```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

income_data = pd.read_csv("Income Data.csv")
income_data.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
columns = ["income", "education", "prestige"]
summary_statistics = pd.DataFrame({
    "Minimum": income_data[columns].apply(np.min),
    "1st Quartile": income_data[columns].apply(lambda values: np.percentile(values, 25)),
    "Median": income_data[columns].apply(np.median),
    "Mean": income_data[columns].apply(np.mean),
    "3rd Quartile": income_data[columns].apply(lambda values: np.percentile(values, 75)),
    "Maximum": income_data[columns].apply(np.max),
}).round(2)

summary_statistics
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(income_data["prestige"], bins=10, color="steelblue", edgecolor="black")
axes[0].set_title("Distribution of Prestige")
axes[0].set_xlabel("Prestige")
axes[0].set_ylabel("Frequency")

axes[1].hist(income_data["education"], bins=10, color="seagreen", edgecolor="black")
axes[1].set_title("Distribution of Education")
axes[1].set_xlabel("Education")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].scatter(income_data["education"], income_data["income"], color="darkorange", alpha=0.75)
axes[0].set_title("Income and Education")
axes[0].set_xlabel("Education")
axes[0].set_ylabel("Income")

axes[1].scatter(income_data["education"], income_data["prestige"], color="purple", alpha=0.75)
axes[1].set_title("Education and Prestige")
axes[1].set_xlabel("Education")
axes[1].set_ylabel("Prestige")

plt.tight_layout()
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
income_variance = np.var(income_data["income"], ddof=1)
income_standard_deviation = np.std(income_data["income"], ddof=1)

print(f"Income variance: {income_variance:.2f}")
print(f"Income standard deviation: {income_standard_deviation:.2f}")
```
