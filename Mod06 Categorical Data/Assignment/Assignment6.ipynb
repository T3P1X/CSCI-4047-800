To measure the psychological state of Americans, a psychology researcher (in 1929) devised
three questions put to 3443 Americans from different walks of life. He asked them whether they
were happy. They could answer only **yes** or **no** to each question. By today’s standards the study
is quite simple, and he did no formal statistical analysis of his data. The frequency data (number
of **yes** and **no** responses within each employment category) from this study are in the file
*Americandata.csv*.

---

You need to use the data and answer the question, that is, **Is the type of profession related to the happiness of American people?**
Report the **χ2** statistic and the *statistical significance of test*, in addition to the contingency table.
Also, explain the results and what they mean. What does the value of **χ2** statistic mean? Is it
good or bad? Is the test statistically significant?

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Rows are happiness responses; columns are professions.
contingency_table = pd.read_csv("Americandata.csv", index_col="Happy")
print("Observed contingency table:")
print(contingency_table.to_string())

# H0: profession and happiness are independent.
# H1: profession and happiness are associated.
alpha = 0.05
chi2, p_value, dof, expected = chi2_contingency(contingency_table)
expected_table = pd.DataFrame(
    expected, index=contingency_table.index, columns=contingency_table.columns
)
print("\nExpected frequencies under independence:")
print(expected_table.round(2).to_string())
print(f"\nTotal responses: {contingency_table.to_numpy().sum()}")
print(f"Chi-square statistic: {chi2:.4f}")
print(f"Degrees of freedom: {dof}")
print(f"p-value: {p_value:.6e}")
print(f"Minimum expected frequency: {expected.min():.2f}")
print(f"Significant at alpha = {alpha}: {p_value < alpha}")
print("\nPercent answering Yes by profession:")
print((100 * contingency_table.loc["Yes"] / contingency_table.sum()).round(2).to_string())
```

### Analysis

I used a **chi-square test of independence** to determine whether profession and happiness are related. The null hypothesis is that profession and happiness are independent; the alternative is that they are associated. I used a significance level of **0.05**. The observed contingency table and expected frequencies are printed above.

The test gives **χ² = 936.1395**, **7 degrees of freedom**, and **p = 7.523435e-198**. The degrees of freedom are (2 − 1)(8 − 1) = 7. Since the p-value is much smaller than 0.05, I reject the null hypothesis. The test is **statistically significant**, providing strong evidence that profession and reported happiness are associated in these data.

The chi-square statistic is the sum of (observed − expected)² / expected across all cells. It measures how far the observed counts differ from the counts expected if profession and happiness were independent. A value near zero indicates close agreement with independence; a larger value indicates greater departure. This large value is evidence of an association, rather than something inherently “good” or “bad,” and it is not a happiness score. Its statistical significance depends on the chi-square distribution and degrees of freedom.

For example, 75.6% of laborers and about 75.7% of physicians answered Yes, compared with 19.5% of students, 11.7% of preachers, and 14.7% of lawyers. Thus the proportion reporting happiness varies substantially across professions. The test does not show that profession causes happiness, nor does the overall test establish which individual differences are statistically significant.

All expected counts exceed 5 (minimum: 17.09), so the expected-frequency condition is satisfied. The analysis assumes independent responses and that each person appears in one cell. Generalizing to all Americans would also require a representative sample; the assignment does not establish how participants were sampled.
