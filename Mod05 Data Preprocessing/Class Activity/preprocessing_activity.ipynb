### 1. A tiny, dirty dataset of quiz scores (out of 100) and letter grades is written below

```python
import pandas as pd
import numpy as np

# A tiny, dirty dataset of quiz scores (out of 100) and letter grades
data = {
    'name':  ['Amir', 'Bella', 'Chen', 'Diego', 'Eve', 'Farah'],
    'score': [88, None, 95, 72, 200, 68],
    'grade': ['A', 'B', 'A', 'c', 'B', 'b'],
}
df = pd.DataFrame(data)
print("Before:")
print(df)
```

### 2. Preprocess
- first step below will be done for you

```python
# 1. NOISE: scores outside 0-100 are impossible. Mark them invalid.
df.loc[(df['score'] < 0) | (df['score'] > 100), 'score'] = np.nan

# 2. MISSING: fill empty scores with the mean of the valid scores.
mean_score = df['score'].mean()
df['score'] = df['score'].fillna(mean_score)

# 3. INCONSISTENT: standardize the letter grades to uppercase.
df['grade'] = df['grade'].str.upper()

# 4. Print the cleaned DataFrame.
print("After:")
print(df)

# 5. Why mark the invalid score as missing first?
# The value 200 would inflate the mean if it were included. Pandas ignores
# NaN when calculating the mean, so we get 80.75 from the valid scores
# (88, 95, 72, 68). Both missing scores are then filled with this value.
```
