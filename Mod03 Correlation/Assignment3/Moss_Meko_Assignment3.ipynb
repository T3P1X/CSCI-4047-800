2. Create a dataframe from the data contained in *income_data.csv* using `pandas`

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("income_data.csv")
print(f"Number of observations: {len(df)}")
print(df.head().to_string(index=False))
```

3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *education*

```python
r_income_education, p_income_education = stats.pearsonr(df["income"], df["education"])
print("Income and education")
print(f"Pearson's r: {r_income_education:.6f}")
print(f"p-value (two-sided): {p_income_education:.6e}")
```

3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *prestige*

```python
r_income_prestige, p_income_prestige = stats.pearsonr(df["income"], df["prestige"])
print("Income and prestige")
print(f"Pearson's r: {r_income_prestige:.6f}")
print(f"p-value (two-sided): {p_income_prestige:.6e}")
```

4. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *education* and *prestige*

```python
r_education_prestige, p_education_prestige = stats.pearsonr(df["education"], df["prestige"])
print("Education and prestige")
print(f"Pearson's r: {r_education_prestige:.6f}")
print(f"p-value (two-sided): {p_education_prestige:.6e}")
```
